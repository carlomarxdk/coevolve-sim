## Load Metadata
Loads paths and basic stats for each completed experiment

```python
# Load all runs
df_runs = load_runs_metadata(Path.cwd().parent / "data"/ "outputs" / "runs")

# Filter for completed and validated runs
df_runs = df_runs[df_runs["completed"] & df_runs["validated"]].copy()

print(f"Total validated runs: {len(df_runs)}")
print("\nRuns by setting:")
print(df_runs.groupby("setting").size())
print("\nRuns by graph type:")
print(df_runs.groupby("graph_type").size())
```

```python
# This dataframe contains all the metadata
# `completed` column indicates whether all the 10 rounds were completed
# `validated` column indicates whether the run passed all validation checks
# (e.g., no missing data, valid graph structure, etc.)
df_runs.head()
```

### Load Agent Data (per run)
To load the `agent_data.json`, you can use `load_agents_data`. 

```python
run_path = df_runs['run_path'].values[1020]
load_agents_data(run_path=run_path).keys()
```

### Load Extended Data (per run)
The `load_run_data` loads:
1. `agents_data.json`
